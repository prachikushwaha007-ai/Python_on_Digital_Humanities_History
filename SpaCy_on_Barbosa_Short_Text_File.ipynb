```python
# To Find People in Barbosa Text File.

people = []
for ent in doc.ents:
  if ent.label_ == "PERSON":
    people.append(ent.text)
print(set(people))
```

```python
nouns = []

for token in sentence:
  if token.pos_ == "NOUN":
    nouns.append(token)

print(nouns)
```

```python
adj = []

for token in sentence:
  if token.pos_ == "ADJ":
    adj.append(token)

print(adj)
```

```python
print(list(doc.noun_chunks))
```

```python
# To Grouping the Noun across Entire Text.
# Find The Specific Noun.

chunks = (list(doc.noun_chunks))

for chunk in chunks:
  if 'Moor' in str(chunk):
    print(chunk)
```

```python
# To Extract Places as geopolitical entity and location

places = []

for ent in doc.ents:
   if ent.label_ in ['GPE','LOC']:
     places.append((ent.text, ent.label_))
print(places)
```

```python
# Lemmatization for find the Roots

for word in sentence:
  if word.pos_ == 'NOUN':
    print(word, word.lemma_)
```
