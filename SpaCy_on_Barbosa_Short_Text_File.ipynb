```python
import spacy
```

```python
# Download large module of spacy.

!python -m spacy download en_core_web_lg

nlp = spacy.load("en_core_web_lg")
```

```python
# Open the File.

with open("/content/drive/MyDrive/Barbosa_Text (1).txt") as f:
  text = f.read()
print(text)
```

```python
#
temp_text = text.replace('\n\n', ' <PARAGRAPH_BREAK> ')

# 2. अब बचे हुए सभी सिंगल \n (जो सेंटेंस के बीच में टूटे हैं) को स्पेस से बदलें
temp_text = text.replace('\n', ' ')

# 3. वापस असली पैराग्राफ ब्रेक रीस्टोर करें
final_text = text.replace(' <PARAGRAPH_BREAK> ', '\n\n')
print(final_text)
```

```python
# To Make a Document of Text.

doc = nlp(text)

# Form The Sentences.

sentences = list(doc.sents)
print(sentences[30])
```

```python
# Apply NER on Document.

ents = list(doc.ents)

for ent in doc.ents:
   print(f"Entity: {ent.text}|Label: {ent.label_}")
```

```python
# To Find People in Barbosa Text File.

ent = list(doc.ents)
people = []
for ent in doc.ents:
  if ent.label_ == "PERSON":
    people.append(ent.text)
print(set(people))
```

```python
sentence = sentences[30]

for token in sentence:
  print(token.text, token.pos_)
```

```python
nouns = []

for token in sentence:
  if token.pos_ == "NOUN":
    nouns.append(token)

print(nouns)
```

```python
adj = []

for token in sentence:
  if token.pos_ == "ADJ":
    adj.append(token)

print(adj)
```

```python
print(list(doc.noun_chunks))
```

```python
# To Grouping the Noun across Entire Text.
# Find The Specific Noun.

chunks = (list(doc.noun_chunks))

for chunk in chunks:
  if 'Moor' in str(chunk):
    print(chunk)
```
