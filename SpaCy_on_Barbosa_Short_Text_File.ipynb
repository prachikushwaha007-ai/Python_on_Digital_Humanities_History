```python
import spacy
```

```python
# Download large module of spacy.

!python -m spacy download en_core_web_lg

nlp = spacy.load("en_core_web_lg")
```

```python
# Open the File.

with open("/content/drive/MyDrive/Text_Barbosa.txt") as f:
  text = f.read()
print(text)
```

```python
# To Make a Document of Text.

doc = nlp(text)

# Form The Sentences.

sentences = list(doc.sents)
print(sentences[10])
```

```python
# Apply NER on Document.

ents = list(doc.ents)

for ent in doc.ents:
   print(f"Entity: {ent.text}|Label: {ent.label_}")
```

```python
# To Find People in Barbosa Text File.

ent = list(doc.ents)
people = []
for ent in doc.ents:
  if ent.label_ == "PERSON":
    people.append(ent.text)
print(set(people))
```
