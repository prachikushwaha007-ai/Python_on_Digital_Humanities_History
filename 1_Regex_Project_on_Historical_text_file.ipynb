```python
# 1.Importing Regex Builtin Module in Python.

import re
```

```python
# 2.Giving Way to Process Text File Stored in Google Drive.

file_path = '/content/drive/MyDrive/freedom of the seas .txt'
```

```python
# 3.To Open Target File Using utf-8 Character Encoding to Properly Handle non-ASCII Characters and Read it as raw_text file.

with open ('/content/drive/MyDrive/freedom of the seas .txt', 'r', encoding = 'utf-8') as f:
     raw_text = f.read()
```

```python
# 4.To Substitute the Special Characters and Symbols from raw_text.

cleaned_text = re.sub(r'[@#$%ˆ&*()\[\]{}]', '', raw_text)
```

```python
# 5.Normalizing Horizontal Spacing by Matching One or More Consecutive Spaces or Tabs and Collapsing it into a Single Standard Space.

cleaned_text = re.sub(r'[ \t]+', ' ', cleaned_text)
```

```python
# 6.Standardizes Paragraph by Identifying Multiple Consecutive Line Breaks and Replacing it with Exactly Two New Lines(\n\n), Creating Paragraph Breaks.

cleaned_text = re.sub(r'\n\s*\n', '\n\n', cleaned_text)
```

```python
# 7.To Remove Page Indicators with Nothing Applying Case Sensitivity Flag (re.IGNORECASE = re.I).

cleaned_text = re.sub(r'page\s+\d+', '', cleaned_text, re.I)
```

```python
# 8.Specifying the Destination File Path where the Cleaned Text Will be Saved.

output_path = '/content/drive/MyDrive/cleaned2_freedom_of_the_seas .txt'
```

```python
# 9.opens a Target File at output_path in Write Mode ('w') Using utf-8 Encoding and Write the Fully Processed cleaned_text String into it.

with open(output_path, 'w', encoding = 'utf-8') as f :
     f.write(cleaned_text)

print(output_path)
```

```python
# EXTRACT DATA FROM CLEANED FILE.

# To Find the Pattern of Specific Word in Cleaned text.

pattern = re.compile(r'\bdutch\b', re.I)

matches = pattern.findall(cleaned_text)
print(matches)


len(matches)
```

```python
# Make a Paragraph belongs to Specific Word ('\n\n').

lines = cleaned_text.split('\n\n')

# via for loop Matching them Having Specific Word.

matching_lines = [line for line in lines if pattern.search(line)]
print(matching_lines)
```

```python
# To Find Years of 16th-18th Century and having paragraphs of these years also.

pattern = re.compile(r'\b(15[0-9]{2}|16[0-9]{2}|17[0-9]{2})\b', re.I)

matches = pattern.findall(cleaned_text)
print(matches)

lines = cleaned_text.split('\n\n')

matching_lines = [ line for line in lines if pattern.search(line)]
print(matching_lines)

len(matches)
```
